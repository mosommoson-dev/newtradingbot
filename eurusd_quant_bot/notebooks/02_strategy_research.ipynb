# 02 — Strategy research

Compare mean-reversion vs trend-following on the same data.

```python
from eurusd_quant_bot.backtest import run_backtest
from eurusd_quant_bot.data import clean_ohlcv, fetch_yfinance
from eurusd_quant_bot.strategy import MeanReversionStrategy, TrendFollowingStrategy

df = clean_ohlcv(fetch_yfinance(granularity='H1', start='2022-01-01'))
for strat_cls in (MeanReversionStrategy, TrendFollowingStrategy):
    s = strat_cls()
    sigs = s.generate_signals(df)
    res = run_backtest(df, sigs)
    print(strat_cls.__name__, res.stats.to_dict())
```
